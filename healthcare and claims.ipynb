```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("Healthcare_Claims_Data_Analyst_Project (2).xlsx")
```

```python
print(df.head())
```

```python
print(df.Shape)
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
status = df["Claim_Status"].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(status.index, status.values)

plt.title("Claims by Status")
plt.xlabel("Claim Status")
plt.ylabel("Number of Claims")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()
```

```python
payers = df.groupby("Payer")["Claim_Amount"].sum()

plt.figure(figsize=(9, 5))

plt.bar(payers.index, payers.values)

plt.title("Total Claim Amount by Payer")
plt.xlabel("Payer")
plt.ylabel("Claim Amount")

plt.xticks(rotation=30)

plt.tight_layout()

plt.show()
```

```python
status = df["Claim_Status"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    status.values,
    labels=status.index,
    autopct="%1.1f%%"
)

plt.title("Claims Status Distribution")

plt.show()
```

```python
denial = df["Denial_Reason"].value_counts()

plt.figure(figsize=(9, 5))

plt.bar(denial.index, denial.values)

plt.title("Claims by Denial Reason")
plt.xlabel("Denial Reason")
plt.ylabel("Number of Claims")

plt.xticks(rotation=30, ha="right")

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Claim_Amount"], bins=20)

plt.title("Distribution of Claim Amount")
plt.xlabel("Claim Amount")
plt.ylabel("Number of Claims")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Claim_Amount"],
    df["Paid_Amount"]
)

plt.title("Claim Amount vs Paid Amount")
plt.xlabel("Claim Amount")
plt.ylabel("Paid Amount")

plt.tight_layout()
plt.show()
```

```python
service = df["Service_Type"].value_counts()

plt.figure(figsize=(9, 5))
plt.bar(service.index, service.values)

plt.title("Claims by Service Type")
plt.xlabel("Service Type")
plt.ylabel("Number of Claims")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```
